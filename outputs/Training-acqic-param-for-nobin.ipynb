```python
# Found in the acqic's grid search
best_params = {
    'learning_rate': 0.05,
    'min_child_samples': 10,
    'min_split_gain': 0.030000000000000002,
    'n_estimators': 10,
    'num_leaves': 53
}
```

```python
D_train = lgb.Dataset(X, y)
```

```python
lgb_params = {
    'boosting_type': 'gbdt',
    'objective': 'binary',
    'device': 'gpu',
}
lgb_params.update(best_params)
eval_res = lgb.cv(
    lgb_params,
    D_train,
    nfold=5,
    metrics='binary_log_loss',
    categorical_feature=cat_feats,
    return_cvbooster=True
)
eval_res
```

```python
models = eval_res['cvbooster']
```

```python
outputs = models.predict(X_test)
preds = [np.where(outputs[i] < 0.5, 0, 1) for i in range(len(outputs))]
y_pred = mode(preds)[0]

res = {
    'txkey': TEST_txkey,
    'label': y_pred
}

pd.DataFrame(res).to_csv('preds.csv', index=False)
```
