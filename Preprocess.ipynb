## 轉換時間格式

```python
def LoctmToSec(loctm):
  sec = loctm % 100
  loctm //= 100
  min = loctm % 100
  loctm //= 100
  hour = loctm % 100
  return 3600 * hour + 60 * min + sec
```

```python
DATA["loctm"] = DATA["loctm"].map(LoctmToSec)
TEST_DATA["loctm"] = TEST_DATA["loctm"].map(LoctmToSec)
```

```python
DATA["chid"] = DATA["chid"].map(CUSTOMER_IDX)
TEST_DATA["chid"] = TEST_DATA["chid"].map(CUSTOMER_IDX)
```

## 補足缺失的幣別

```python
# 實付金額 flam1
# 消費地金額 csmam
money_cols = ["flam1", "csmcu", "csmam"]
money_train = DATA[~np.isnan(DATA["csmcu"]) & (DATA["csmam"] > 0)][money_cols]
money_test = TEST_DATA[~np.isnan(TEST_DATA["csmcu"]) & (TEST_DATA["csmam"] > 0)][money_cols]

money_data = pd.concat((money_train, money_test))
money_data["exchange_rate"] = money_data["flam1"] / money_data["csmam"]
grouped = money_data[["csmcu", "exchange_rate"]].groupby("csmcu")
avg_exchange_rate_table = grouped.mean()
```

```python
def InferCsmcu(exchange_rate):
    return (avg_exchange_rate_table['exchange_rate'] - exchange_rate).abs().idxmin()
```

```python
target_train = np.isnan(DATA["csmcu"]) & ~np.isnan(DATA["flam1"]) & (DATA["csmam"] > 0)
target_test = np.isnan(TEST_DATA["csmcu"]) & ~np.isnan(TEST_DATA["flam1"]) & (TEST_DATA["csmam"] > 0)
money_train = DATA[target_train]
money_test = TEST_DATA[target_test]

DATA.loc[target_train, 'csmcu'] = (money_train["flam1"] / money_train["csmam"]).map(InferCsmcu)
TEST_DATA.loc[target_test, 'csmcu'] = (money_test["flam1"] / money_test["csmam"]).map(InferCsmcu)
```

```python
DATA = DATA.drop(columns='csmam')
TEST_DATA = TEST_DATA.drop(columns='csmam')
```

## 新增實付金額 / 交易金額 Ratio

```python
# 實付金額 flam1
# 交易金額 conam
DATA["fc_ratio"] = (DATA["flam1"] / DATA["conam"])
DATA.loc[np.isnan(DATA['fc_ratio']), 'fc_ratio'] = 1.0
DATA = DATA.drop(columns='flam1')

TEST_DATA["fc_ratio"] = (TEST_DATA["flam1"] / TEST_DATA["conam"])
TEST_DATA.loc[np.isnan(TEST_DATA['fc_ratio']), 'fc_ratio'] = 1.0
TEST_DATA = TEST_DATA.drop(columns='flam1')
```

## 補足缺乏的國別資料
- 無 stocn 但有 scity 的資料，有機會使用現有的資料推出 stocn
- 有 stocn 無 scity 的資料和兩者皆無的資料就幫不上忙了

不過只要是台灣以外的地區我們都只看國別，所以只有 stocn = 0 但 scity = NaN 的 case 會需要另外處理

```python
loc_cols = ['stocn', 'scity']
LOC_DATA = pd.concat((DATA[loc_cols], TEST_DATA[loc_cols]), ignore_index=True)
```

```python
dedup_stocn_scity_pairs = LOC_DATA[~np.isnan(LOC_DATA['stocn']) & ~np.isnan(LOC_DATA['scity'])].drop_duplicates()
scity_to_stocn_map = dedup_stocn_scity_pairs.set_index('scity').to_dict()['stocn']

nan_rows = np.isnan(LOC_DATA['stocn'])
# The following is semantically equal to LOC_DATA[nan_rows]['stocn'] = ...
# However, it's safer and doesn't cause undefined behavior
LOC_DATA.loc[nan_rows, 'stocn'] = LOC_DATA.loc[nan_rows, 'scity'].map(scity_to_stocn_map)
```

## 轉換地點類別
NOTE: 需要重新編號，避免 scity 和 stocn 衝到
- stocn = 0, scity != NaN
    - scity (re-labeled scity)
- stocn = 0, scity = NaN
    - new category (number of unique scity)
    - 事實上在 training data 中，補足資料後就沒有這個 case 了
- stocn > 0
    - stocn (stocn + number of unique scity)
- stocn = NaN
    - NaN

```python
_, idx = pd.factorize(LOC_DATA['scity'][LOC_DATA['stocn'] == 0])
# idx contains index of unique values excluding NaN
scity_unique_num = idx.shape[0]

train_tw_rows = DATA['stocn'] == 0
test_tw_rows = TEST_DATA['stocn'] == 0
scity_to_id = dict([(scity, id) for id, scity in enumerate(idx)])

DATA.loc[train_tw_rows, 'region'] = DATA.loc[train_tw_rows, 'scity'].map(scity_to_id)
TEST_DATA.loc[test_tw_rows, 'region'] = TEST_DATA.loc[test_tw_rows, 'scity'].map(scity_to_id)

DATA.loc[train_tw_rows & np.isnan(DATA['scity']), 'region'] = scity_unique_num
TEST_DATA.loc[test_tw_rows & np.isnan(TEST_DATA['scity']), 'region'] = scity_unique_num

DATA.loc[~train_tw_rows, 'region'] = DATA.loc[~train_tw_rows, 'stocn'] + scity_unique_num
TEST_DATA.loc[~test_tw_rows, 'region'] = TEST_DATA.loc[~test_tw_rows, 'stocn'] + scity_unique_num
```

```python
DATA = DATA.drop(columns=loc_cols)
TEST_DATA = TEST_DATA.drop(columns=loc_cols)
```

## 新增特定欄位中各類別的 Fraud Ratio
除了自己分 bins 外，也可以放 numerical features 就好，讓 LightGBM 去試

```python
FRAUD_DATA = DATA[DATA['label'] == 1]
```

```python
# fraud ratio of each category in specified feature
def FraudRatio(column):
    data = DATA[column].value_counts()
    idx = data.index
    count = data.values

    ratio = [
        (FRAUD_DATA[column] == i).sum() / c
        # if c > 20 else np.nan
        for i, c in zip(idx, count)
    ]

    return pd.Series(ratio, index=idx).to_dict()
```

```python
def FindBin(ratio, boundaries):
    if np.isnan(ratio):
        return np.nan
    # first bin is closed at both boundaries
    # other bins are open at left boundary and closed at right boundary
    if ratio == 0:
        return 1
    return np.searchsorted(boundaries, ratio, side="left")
```

```python
mcc_fraud_ratio_map = FraudRatio("mcc")

mcc_fraud_ratio = DATA["mcc"].map(mcc_fraud_ratio_map)
DATA["mcc_fr"] = mcc_fraud_ratio

mcc_fraud_ratio = TEST_DATA["mcc"].map(mcc_fraud_ratio_map)
TEST_DATA["mcc_fr"] = mcc_fraud_ratio
```

```python
region_fraud_ratio_map = FraudRatio("region")

region_fraud_ratio = DATA["region"].map(region_fraud_ratio_map)
DATA["region_fr"] = region_fraud_ratio

region_fraud_ratio = TEST_DATA["region"].map(region_fraud_ratio_map)
TEST_DATA["region_fr"] = region_fraud_ratio
```

```python
chid_fraud_ratio_map = FraudRatio("chid")

chid_fraud_ratio = DATA["chid"].map(chid_fraud_ratio_map)
DATA["chid_fr"] = chid_fraud_ratio

chid_fraud_ratio = TEST_DATA["chid"].map(chid_fraud_ratio_map)
TEST_DATA["chid_fr"] = chid_fraud_ratio
```

```python
DATA = DATA.drop(columns=["mcc", "region", "chid"])
TEST_DATA = TEST_DATA.drop(columns=["mcc", "region", "chid"])
```

## One-Hot Encoding

```python
cat_feats = [
    'contp',
    'etymd',
    'ecfg',
    'bnsfg',
    'stscd',
    'ovrlt',
    'flbmk',
    'hcefg',
    'csmcu',
    'flg_3dsmk',
]
```

## Deal with NaN in Numerical Features
NN 不接受 nan 的輸入

```python
for col in DATA.columns:
    nan_count = DATA[col].isna().sum()
    if nan_count > 0:
        print(col, nan_count)
```

```python
for col in TEST_DATA.columns:
    nan_count = TEST_DATA[col].isna().sum()
    if nan_count > 0:
        print(col, nan_count)
```

```python
imputer = SimpleImputer(missing_values=np.nan, strategy="mean")

def ImputeData(data):
    imputed_data = imputer.fit_transform(data)
    imputed_data = pd.DataFrame(imputed_data, columns=data.columns)
    return imputed_data
```

```python
DATA = ImputeData(DATA)

test_txkey = TEST_DATA["txkey"]
TEST_DATA = ImputeData(TEST_DATA.iloc[:, 1:])
TEST_DATA["txkey"] = test_txkey
```
